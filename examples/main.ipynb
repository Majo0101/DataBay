## Reporting — `databay.reporting.plot`

Vizualizácie výsledkov porovnávania datasetov pomocou Plotly.  
K dispozícii sú tri funkcie:
- `plot_match_percentage` — horizontálny bar chart s farebným kódovaním kvality
- `plot_match_percentage_sorted` — rovnaká vizualizácia, automaticky zoradená
- `plot_overlap` — stacked bar chart ukazujúci prekryv dvoch datasetov

### `plot_match_percentage` — bar chart kvality stĺpcov

```python
from databay.reporting.plot import plot_match_percentage

# Simulovane vysledky porovnania stlpcov PROD vs TEST
match_data = {
    "customer_id": 100.0,
    "name": 99.1,
    "email": 97.4,
    "balance": 85.3,
    "status": 62.0,
    "created_at": 100.0,
    "phone": 43.7,
    "address": 91.2,
}

fig = plot_match_percentage(
    match_data,
    title="Column Match: PROD vs TEST",
    show_reference_lines=True,   # ciary na 95 % a 99 %
    sort=False,
)
fig.show()
```

### `plot_match_percentage_sorted` — zoradené podľa match %

```python
from databay.reporting.plot import plot_match_percentage_sorted

# Rovnaky slovnik ? tentokrat zoradene od najhorsieho (ascending=True = najnizsi match prvy)
fig = plot_match_percentage_sorted(
    match_data,
    title="Column Match: PROD vs TEST (Sorted - najhorsie navrchu)",
    ascending=True,
    show_reference_lines=True,
)
fig.show()
```

### `plot_overlap` - funnel overlap Dataset A vs Dataset B

2 bary: Dataset A a Dataset B.
- zelena = overlap
- zlta = iba v Dataset A
- cervena = iba v Dataset B


```python
from databay.reporting.plot import plot_overlap

overlap_data = {
    "left_count": 12000,   # Dataset A total
    "right_count": 11000,  # Dataset B total
    "overlap_count": 9800, # rows in both A and B
}

fig = plot_overlap(
    overlap_data,
    title="Dataset overlap funnel: A vs B",
    left_name="Dataset A",
    right_name="Dataset B",
    overlap_color="#43a047",  # green
    left_only_color="#f9a825", # yellow
    right_only_color="#e53935",# red
)
fig.show()
```
