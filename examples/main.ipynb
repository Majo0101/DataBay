# DataBay
Choose direct Spark or Spark Connect before starting your Python kernel.

```python
from databay.runtime import spark_direct, spark_connect
from databay import null_rate, duplicate_check

# Choose one mode per kernel.
mode = "direct"
if mode == "direct":
    spark = spark_direct()
else:
    spark = spark_connect()
df = spark.createDataFrame([(1, "Ana"), (2, None), (2, "Peter")], "id long, name string")
null_rate(df).show()
duplicate_check(df, cols=["id"]).show()
```
